```python
%sql

drop table if exists procat.proschema.gold_orders;
create table procat.proschema.gold_orders (
    order_id string,
    order_date string,
    customer_id string,
    region string,
    product_name string,
    unit_price decimal(12,2),
    quantity integer,
    order_amount decimal(14,2),
    order_year int,
    order_month int
)
using delta
partitioned by(order_year,order_month)
```

```python
%sql

insert overwrite table procat.proschema.gold_orders
partition (order_year,order_month)
select 
  order_id,
  order_date,
  customer_id,
  region,
  product_name,
  unit_price,
  quantity,
  order_amount,

  year(order_date) as order_year,
  month(order_date) as order_month

from procat.proschema.silver_orders;
```
